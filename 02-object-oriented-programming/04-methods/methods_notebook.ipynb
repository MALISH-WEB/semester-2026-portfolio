```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def introduce(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")
P1 = Student("Alice", 20)
P1.introduce()
```

```python
# Function 

class Phone:
    def __init__(self, brand, storage_capacity):
        self.brand = brand
        self.storage_capacity = storage_capacity
    
    def display_info(self):
        print(f"Brand: {self.brand}")
        print(f"Storage Capacity: {self.storage_capacity} GB")


# Simulate making a call

    def make_call(self, number):
        print(f"Calling {number} from {self.brand} phone...")

Phone1 = Phone("Apple", 128)
Phone2 = Phone("Samsung", 256)
Phone1.display_info()
Phone2.display_info()
Phone1.make_call("078-456-7890")
Phone2.make_call("078-765-4321")
```

```python
class Bank:
    def __init__(self, account_number, holder_name, balance):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Insufficient funds.")

Bank1 = Bank("0983416780", "John Doe", 10000)
Bank1.deposit(50000)
Bank1.deposit(70000)
Bank1.withdraw(200)
```
