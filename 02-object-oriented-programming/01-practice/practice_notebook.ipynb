```python
class Student:

    def __init__(self, reg_number, name, age, gender, phone_number, email, programme):
        self.reg_number = reg_number
        self.name = name
        self.age = age
        self.gender = gender
        self.phone_number = phone_number
        self.email = email
        self.programme = programme

student1 = Student("REG123", "Alice", 20, "Female", "1234567890", "alice@example.com", "Computer Science")
student2 = Student("REG456", "Bob", 22, "Male", "0987654321", "bob@example.com", "Information Technology")

print("\nStudent Details")
print(student1.reg_number)
print(student1.name)
print(student1.age)
print(student1.gender)
print(student1.phone_number)
print(student1.email)
print(student1.programme)
```

```python
class student:
    def __init__(self,name, age, gender, phone_number, email, programme):
        self.name = name
        self.age = age
        self.gender = gender
        self.phone_number = phone_number
        self.email = email
        self.programme = programme

student1 = student("Alice", 20, "Female", "1234567890", "alice@example.com", "Computer Science")
student2 = student("Bob", 22, "Male", "0987654321", "bob@example.com", "Information Technology")
student3 = student("Charlie", 19, "Male", "1111111111", "charlie@example.com", "Mathematics")

print(student1.name, student1.age, student1.gender, student1.phone_number, student1.email, student1.programme)
print(student2.name, student2.age, student2.gender, student2.phone_number, student2.email, student2.programme)
print(student3.name, student3.age, student3.gender, student3.phone_number, student3.email, student3.programme)
```
