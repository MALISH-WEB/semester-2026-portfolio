```python
# Encapsulation

class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.__account_number = account_number  # Private attribute
        self.account_holder = account_holder  # Public attribute
        self.__balance = balance  # Private attribute


    def deposit(self, amount):
        if amount >0:
            self.__balance += amount
            print("Deposit successful. New balance: ${}".format(self.__balance))
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal successful. New balance: ${}".format(self.__balance))
        else:
            print("Invalid withdrawal amount. Please check your balance.")

    def get_balance(self):
        return self.__balance  # Public method to access private attribute
```

```python
class Account:
    def __init__(self, holder_name, balance, amount):
        self.holder_name = holder_name
        self.__balance = balance  # Private attribute
        self.__amount = amount  # Private attribute

    def get_balance(self):
        return self.__balance  # Public method to access private attribute

    def get_amount(self):
        return self.get_amount

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful. New balance: ${}".format(self.__balance))
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
            if amount > 0 and amount <= self.__balance:
                self.__balance -= amount
                print("Withdrawal successful. New balance: ${}".format(self.__balance))
            else:
                print("Invalid withdrawal amount. Please check your balance.")

    def check_balance(self):
        return self.__balance  # Public method to access private attribute

account1 = Account("Malish", 50000000, 6000000)

print(account1.get_balance())
account1.deposit(67709)
account1.withdraw(5000000)
print("Current Balance:", account1.check_balance())

    
```
